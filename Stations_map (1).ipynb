```python
import folium

# Define the map center (e.g., the average latitude and longitude)
map_center = [df['latitude'].mean(), df['longitude'].mean()]
m = folium.Map(location=map_center, zoom_start=13)

# Add markers for each station
for _, row in df.iterrows():
    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=f"<b>Station:</b> {row['name']}",
        tooltip=f"{row['name']}"
    ).add_to(m)

# Save the map to an HTML file
m.save("stations_map.html")

# Optional: Display the map if using Jupyter Notebook
m
```

```python
# Replace 'path_to_your_file.xlsx' with the path to your Excel file
file_path = "C:\\Users\\baraa\\Downloads\\bquxjob_7862c34f_1949876f6b9.csv"

# Load the Excel file into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure the data is loaded correctly
print(df.head())
```

```python
import folium

# Define the map center (use the average latitude and longitude)
map_center = [df['latitude'].mean(), df['longitude'].mean()]
m = folium.Map(location=map_center, zoom_start=13)

# Add markers for each station based on status
for _, row in df.iterrows():
    # Set marker color based on the status
    if row['status'] == 'available':
        color = 'blue'
    elif row['status'] == 'closed':
        color = 'red'
    else:
        color = 'gray'  # Default color for unknown status
    
    # Create popup content
    popup_text = f"""
    <b>Station:</b> {row['name']}<br>
    <b>Bikes Available:</b> {row['num_of_bikes']}<br>
    <b>Status:</b> {row['status']}
    """
    
    # Add the marker to the map
    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=popup_text,
        tooltip=f"{row['name']} ({row['status']})",
        icon=folium.Icon(color=color)  # Set marker color
    ).add_to(m)

# Save the map to an HTML file
m.save("stations_map_with_status.html")

# Optional: Display the map in Jupyter Notebook
m
```
